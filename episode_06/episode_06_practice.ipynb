````python
def check(name, expected, tolerance=0.001):
    """Compare a variable you created with the value the exercise expects."""
    if name not in globals():
        print(f"⚠️  Create a variable called {name} first.")
        return
    value = globals()[name]
    if isinstance(expected, float) and isinstance(value, (int, float)):
        ok = abs(value - expected) <= tolerance
    else:
        ok = value == expected
    print(f"✅ {name} = {value!r}" if ok else f"❌ {name} is {value!r}, but it should be {expected!r}")


print("Ready! Run the cells below one at a time with Shift + Enter.")
````

## 1. print() with sep and end

`sep` is the text between values (default: one space). `end` is the text after the last value (default: a new line).
Because `print` shows text instead of returning it, these exercises build the same text with a variable.

````python
print("Rs", 1500, "paid")
print("Rs", 1500, "paid", sep="")
print("2026", "11", "03", sep="-")
print("Name:", end=" ")
print("Asha")
````

## 2. Escape characters

An **escape character** is a backslash: `\n` new line, `\t` tab, `\"` a quote mark, `\\` one backslash.

````python
print("Line one\nLine two")
print("Name\tBalance")
print("She said \"hi\"")
````

## 3. f-strings and format specifiers

An **f-string** puts values inside curly braces. A **format specifier** after a colon controls how a value is shown:
`.2f` is two decimals and `,` adds thousands separators.

````python
deposit = 1500.5
print(deposit)
print(f"{deposit:.2f}")
print(f"{deposit:,.2f}")
````
