## 1. Concatenation and len()

Plus joins text. Joining text is called **concatenation**. `len()` counts characters, spaces included.

````python
first = "Asha"
last = "Rao"
print(first + last)
print(first + " " + last)
print(len("Asha Rao"))
````

## 2. Index and slices

An **index** is a position, counted from zero. A minus index counts from the end.
A **slice** `word[start:stop]` stops before the stop number.

````python
word = "Savings"
print(word[0])
print(word[-1])
print(word[0:3])
print(word[3:])
````

## 3. Methods

A **method** is an action that belongs to a value: `value.method()`.
`upper()` makes capitals. `strip()` removes spaces from both ends.
A method gives back **new** text, so save it in a variable.

````python
name = "  asha rao  "
clean = name.strip().upper()
print(clean)
print(name)
````
