````python
def check(name, expected, tolerance=0.001):
    """Compare a variable you created with the value the exercise expects."""
    if name not in globals():
        print(f"⚠️  Create a variable called {name} first.")
        return
    value = globals()[name]
    if isinstance(expected, float) and isinstance(value, (int, float)):
        ok = abs(value - expected) <= tolerance
    else:
        ok = value == expected
    print(f"✅ {name} = {value!r}" if ok else f"❌ {name} is {value!r}, but it should be {expected!r}")


print("Ready! Run the cells below one at a time with Shift + Enter.")
````

## 4. Run the whole program

Run this cell and type `5000`, then `6`, then `12`. You should see a balance of 61,677.81.
Then try your own numbers. Predict the balance first!

````python
monthly = float(input("Monthly savings: "))
annual_rate = float(input("Yearly interest rate (%): "))
months = int(input("Months: "))

rate = annual_rate / 12 / 100
saved = monthly * months
balance = monthly * ((1 + rate) ** months - 1) / rate
interest = balance - saved

print(f"Saved: {saved:,.2f}")
print(f"Balance: {balance:,.2f}")
print(f"Interest: {interest:,.2f}")
````
