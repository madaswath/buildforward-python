````python
def check(name, expected, tolerance=0.001):
    """Compare a variable you created with the value the exercise expects."""
    if name not in globals():
        print(f"⚠️  Create a variable called {name} first.")
        return
    value = globals()[name]
    if isinstance(expected, float) and isinstance(value, (int, float)):
        ok = abs(value - expected) <= tolerance
    else:
        ok = value == expected
    print(f"✅ {name} = {value!r}" if ok else f"❌ {name} is {value!r}, but it should be {expected!r}")


print("Ready! Run the cells below one at a time with Shift + Enter.")
````

## 1. Three conversion tools

`int()` makes a whole number and **chops** decimals. `float()` makes a decimal number. `str()` makes text.

````python
print(int("500") + 500)
print(int(3.9), int(-3.9), round(3.9))
print(float("1500.5") + 1)
print("Balance: " + str(5000))
````

## 2. Reading errors

Read a **traceback** from the bottom up: the last line says what went wrong, the line above shows the code,
and the `File ..., line N` line says where.

Error 1 (**TypeError**): `print("Total: " + 500)` stops with
`TypeError: can only concatenate str (not "int") to str`. Fix: `"Total: " + str(500)`.

Error 2 (**TypeError**): `print(500 + "500")` stops with
`TypeError: unsupported operand type(s) for +: 'int' and 'str'`. Fix: `500 + int("500")`.

Error 3 (**ValueError**): `int("1500.5")` stops with
`ValueError: invalid literal for int() with base 10: '1500.5'`. Fix: `float("1500.5")`.

**3. What does a ValueError mean?**
