**Run this setup cell once.** It gives you the `check` helper used by the exercises.

````python
def check(name, expected, tolerance=0.001):
    """Compare a variable you created with the value the exercise expects."""
    if name not in globals():
        print(f"⚠️  Create a variable called {name} first.")
        return
    value = globals()[name]
    if isinstance(expected, float) and isinstance(value, (int, float)):
        ok = abs(value - expected) <= tolerance
    else:
        ok = value == expected
    print(f"✅ {name} = {value!r}" if ok else f"❌ {name} is {value!r}, but it should be {expected!r}")


print("Ready! Run the cells below one at a time with Shift + Enter.")
````

## 3. How Python runs your code

In the video, CPython turns your source code into **bytecode**: small, simple steps. Python can show them to you. Run this and compare each line with the video.

````python
import dis

dis.dis("new_balance = balance + deposit")
````

You will see steps like `LOAD_NAME` (fetch a variable), `BINARY_OP` (do the addition) and `STORE_NAME` (save the result). The exact numbers on the left can differ between Python versions. That is fine.

## 4. Read your first errors

Everyone meets **error messages**. They are Python telling you what went wrong. Read them **from the bottom**: the last line names the kind of mistake.

These cells make the mistake on purpose, but catch the error so you can read it safely.

**Mistake 1: names are case sensitive.** We created `balance`, then used `Balance`.

````python
try:
    balance = 5000
    print(Balance)
except NameError as error:
    print("Python says:", error)
````

**Mistake 2: indentation.** The `print` must sit inside the `if` block. Python reports this before running, so we compile the text to show the message.

````python
source = 'balance = 5000\nif balance > 0:\nprint(balance)\n'
try:
    compile(source, "mistake.py", "exec")
except IndentationError as error:
    print("Python says:", error)
````

**Mistake 3: quotes must match.** The string below never closes.

````python
source = 'customer = "Asha\nprint(customer)\n'
try:
    compile(source, "mistake.py", "exec")
except SyntaxError as error:
    print("Python says:", error)
````
