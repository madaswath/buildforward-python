## 2. Floor division and modulo

**Floor division** `//` keeps the whole part. **Modulo** `%` keeps the remainder.
They are partners.

````python
print(7 / 2)
print(7 // 2)
print(7 % 2)
paise = 15050
print("Rupees:", paise // 100)
print("Paise:", paise % 100)
````

## 3. Exponents, precedence, round()

Two stars `**` raise to a power. **Precedence** decides order: brackets, powers, times/divide, then plus/minus.
`round(2.675, 2)` is `2.67` because the float is stored just below 2.675.

````python
print(2 ** 10)
print(2 + 3 * 4)
print((2 + 3) * 4)
print(round(2.675, 2))
````
