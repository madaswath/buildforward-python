## 1. if and else

A **condition** is an expression that gives `True` or `False`. `if` runs its **block** (the indented lines) only when the condition is `True`.
`else` runs when it is `False`.

````python
balance = 5000
withdrawal = 3000
if balance >= withdrawal:
    print("Approved")
else:
    print("Declined")
````

## 4. Try the decision program

Run it three times: 3000, 7500.5 and -200. Predict each answer first.

````python
balance = 5000.0
amount = float(input("Withdraw: "))

if amount <= 0:
    print("Invalid amount")
elif amount > balance:
    print("Declined: not enough money")
else:
    new_balance = balance - amount
    print("Approved")
    print(f"New balance: {new_balance:,.2f}")
````

**5. Why does `if amount >= 5000` before `elif amount >= 20000` give a wrong answer for 25000?**

## What you learned

- `if` runs a block only when its condition is `True`
- Indented lines form a block (four spaces)
- `else` covers the other case; `elif` adds more cases, checked from the top
- Order and indentation decide everything
