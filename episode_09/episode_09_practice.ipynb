One equals sign (`=`) stores a value. Two (`==`) ask a question.

**Run this setup cell once.** It gives you the `check` helper used by the exercises.

````python
def check(name, expected, tolerance=0.001):
    """Compare a variable you created with the value the exercise expects."""
    if name not in globals():
        print(f"⚠️  Create a variable called {name} first.")
        return
    value = globals()[name]
    if isinstance(expected, float) and isinstance(value, (int, float)):
        ok = abs(value - expected) <= tolerance
    else:
        ok = value == expected
    print(f"✅ {name} = {value!r}" if ok else f"❌ {name} is {value!r}, but it should be {expected!r}")


print("Ready! Run the cells below one at a time with Shift + Enter.")
````

## 1. Comparison operators

A **boolean** is `True` or `False`. A **comparison operator** compares two values: `>`, `<`, `>=`, `<=`, `==`, `!=`.

````python
balance = 5000
print(balance > 3000)
print(balance < 3000)
print(balance >= 5000)
print(balance == 4000)
print(balance != 0)
````

## 4. Truthiness

**Truthiness**: `bool(value)` shows how Python treats a value. Zero and empty are `False`; most other values are `True`.
The text `"0"` is not empty, so `bool("0")` is `True`.

````python
print(bool(5000), bool(0), bool(0.0))
print(bool("Asha"), bool(""), bool("0"))
````
