````python
def check(name, expected, tolerance=0.001):
    """Compare a variable you created with the value the exercise expects."""
    if name not in globals():
        print(f"⚠️  Create a variable called {name} first.")
        return
    value = globals()[name]
    if isinstance(expected, float) and isinstance(value, (int, float)):
        ok = abs(value - expected) <= tolerance
    else:
        ok = value == expected
    print(f"✅ {name} = {value!r}" if ok else f"❌ {name} is {value!r}, but it should be {expected!r}")


print("Ready! Run the cells below one at a time with Shift + Enter.")
````

## 2. The kind of value: type()

Four basic types: **int** (whole number), **float** (has a decimal point), **str** (text in quotes), **bool** (`True` or `False`).

````python
print(type(5000))
print(type(6.5))
print(type("Asha"))
print(type(True))
print(type("5000"))
````

## 3. Naming rules

Rules: letters, digits, and underscores only. Never start with a digit. No spaces. Capital letters matter.
Habit: lowercase words joined by underscores, and names that say what the value is.

Python refuses these two lines:

```
2nd_account = 100
my balance = 100
```

## 4. The 0.1 + 0.2 puzzle

Floats are stored as close values, not exact ones. That is why `0.1 + 0.2` prints a long tail.
Habits: do not expect a float to be exact, round when you show an answer (Episode 4), and keep money in whole units.

````python
print(0.1 + 0.2)
````
